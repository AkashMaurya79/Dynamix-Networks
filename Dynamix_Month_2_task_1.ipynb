```python
# Task 4:- Data Visualization
```

```python
df['status'].value_counts()
```

```python
plt.figure(figsize=(12, 7))
sns.countplot(x='status', data=df, palette='viridis')
plt.title('Distribution of Mental Health Statuses', fontsize=16)
plt.xlabel('Status', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

```python
df['statement'] = df['statement'].astype(str)
df['statement_length'] = df['statement'].apply(len)
df['statement_length'].value_counts()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['statement_length'], bins=50, kde=True, color='skyblue')
plt.title('Distribution of Statement Lengths', fontsize=16)
plt.xlabel('Statement Length', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.tight_layout()
plt.show()
```
