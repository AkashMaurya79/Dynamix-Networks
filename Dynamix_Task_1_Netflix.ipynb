```python
# Task 3:- Exploratory Data Analysis (EDA)
```

```python
sns.countplot(data=df, x='type')
plt.title('Distribution of Content Types')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
country_counts_by_type = df.groupby('type')['country'].nunique().reset_index()

sns.barplot(data=country_counts_by_type, x='type', y='country')
plt.title('Number of Unique Countries per Content Type')
plt.xlabel('Type')
plt.ylabel('Number of Unique Countries')
plt.show()
```

```python
# Task 4:- Visualizations
```

```python
# Extract year and month from 'date_added'
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
# Plot content added over the years
plt.figure(figsize=(12, 6))
sns.countplot(x='year_added', data=df, hue='year_added', palette='coolwarm', legend=False)
plt.title('Content Added Over Time')
plt.xlabel('Year')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
top_directors = df['director'].value_counts().head(10)
top_directors
```

```python
# Count titles by director
top_directors = df['director'].value_counts().head(10)
# Plot top directors
plt.figure(figsize=(15, 7))
sns.barplot(x=top_directors.index,y=top_directors,palette='Blues_d')
plt.title('Top 10 Directors with the Most Titles')
plt.xticks(rotation=45)
plt.ylabel('Number of Titles')
plt.xlabel('Director')
plt.show()
```

```python
freq=df['type'].value_counts()
fig,axes=plt.subplots(1,2,figsize=(8,4))
sns.countplot(df,x=df['type'],ax=axes[0])
plt.pie(freq,labels=['Movie','TVShow'],autopct='%.0f%%')
plt.suptitle('Total Contenton Netflix',fontsize=20)
```

```python
df['rating'].value_counts()
```

```python
ratings=df['rating'].value_counts().reset_index().sort_values(by='count', ascending=False)
plt.pie(ratings['count'][:8], labels=ratings['rating'][:8],autopct='%.0f%%')
plt.suptitle('Rating on Netflix', fontsize=20)
```
