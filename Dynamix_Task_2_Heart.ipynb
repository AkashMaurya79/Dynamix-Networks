```python
# Task 3:- Exploratory Data Analysis (EDA)
```

```python
df.sex.value_counts()
```

```python
feg,axes=plt.subplots(1,2,figsize=(8,6))
sns.countplot(data=df,x=df['sex'],ax=axes[0])
axes[0].set_title("Count of Male and Female")
plt.pie(df.sex.value_counts(),labels=df.sex.value_counts().index,autopct="%1.2f%%")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.title("Count of Male and Female")
plt.show()
```

```python
df.chest_pain_type.value_counts()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=df,x=df['chest_pain_type'],palette='viridis')
plt.xlabel("Chest Pain Type")
plt.ylabel("Count")
plt.title("Count of Chest Pain Type")
plt.show()
```

```python
# Observation of Task 3:-
# In this task , I Observe the gender contribution of that dataset that is 206 are males and 96 are females & in the percentage ratio the males are 68.21% and 31.79% are females.
# And also I observe the total patients of the different types of chest pain and found the maximum patients are trouble with Typical angina.
```

```python
# Task 4:- Visualizations
```

```python
df.slope.value_counts()
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=df,x=df['slope'],palette='viridis')
plt.xlabel("Slope")
plt.ylabel("Count")
plt.title("Count of Slope")
plt.show()
```

```python
df.target.value_counts()
```

```python
df.fasting_blood_sugar.value_counts()
```

```python
plt.figure(figsize=(8,6))
fasting_sugar_counts = df['fasting_blood_sugar'].value_counts()
plt.pie(fasting_sugar_counts, labels=fasting_sugar_counts.index, autopct='%.0f%%')
plt.title("Count of Fasting Blood Sugar")
plt.show()
```

```python
ecg_counts=df.rest_ecg.value_counts()
ecg_counts
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=df,x='rest_ecg',palette='viridis')
plt.xlabel("Rest ECG")
plt.ylabel("Count")
plt.title("Count of Rest ECG")
plt.show()
```

```python
exercise_accor_gender=df.groupby('exercise_induced_angina')['sex'].value_counts()
exercise_accor_gender
```

```python
plt.figure(figsize=(8,6))
sns.countplot(data=df,x='exercise_induced_angina',hue='sex',palette='viridis')
plt.xlabel("Exercise Induced Angina with gender")
plt.ylabel("Count")
plt.title("Count of Exercise Induced Angina with gender")
plt.show()
```
