```python
# Task 4:- Data Visualization:-
```

```python
top_city = df['City'].value_counts().head(10)
top_city
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=top_city.index,y=top_city.values,palette='viridis')
plt.title('Top 10 Cities')
plt.xlabel('City')
plt.ylabel('Count')
plt.show()
```

```python
payment_method = df['Payment_Method'].value_counts()
payment_method
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x=payment_method.index,y=payment_method.values,palette='viridis')
plt.title('Payment Method Distribution')
plt.xlabel('Payment Method')
plt.ylabel('Count')
plt.show()
```

```python
device_type = df['Device_Type'].value_counts()
device_type
```

```python
plt.figure(figsize=(10,5))
plt.pie(device_type,labels=device_type.index,autopct='%1.1f%%',colors=sns.color_palette('viridis'))
plt.title('Device Type Distribution')
plt.xlabel('Device Type')
plt.ylabel('Count')
plt.show()
```

```python
top_city_with_payment_method = df.groupby(['Payment_Method','City']).size().reset_index(name='Count').sort_values(by='Count',ascending=False).head(10)
top_city_with_payment_method
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='City', y='Count', hue='Payment_Method', data=top_city_with_payment_method, palette='viridis')
plt.title('Top 10 Cities with Payment Method')
plt.xlabel('City')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_Payment_with_device_type = df.groupby(['Payment_Method','Device_Type']).size().reset_index(name='Count')
top_Payment_with_device_type
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Payment_Method', y='Count', hue='Device_Type', data=top_Payment_with_device_type, palette='viridis')
plt.title('Payment Method with Device Type')
plt.xlabel('Payment Method')
plt.ylabel('Count')
plt.show()
```

```python
rating_with_gender = df.groupby(['Gender','Customer_Rating']).size().reset_index(name='Count')
rating_with_gender
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Gender', y='Count', hue='Customer_Rating', data=rating_with_gender, palette='viridis')
plt.title('Customer Rating with Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
gender_count= df['Gender'].value_counts()
gender_count
```

```python
plt.figure(figsize=(10,5))
plt.pie(gender_count,labels=gender_count.index,autopct='%1.1f%%',colors=sns.color_palette('viridis'))
plt.title('Gender Distribution')
plt.show()
```
